```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
DATA_data = pd.read_csv("THORACIC_DATA_2025.DAT", delimiter="\t", header=None)
```

```python
DATA_cn = pd.read_excel(r"C:\Users\Omar\Documents\HeartTX\column_names_2025.xlsx", header=None)[0].tolist()
```

```python
DATA_data.columns = DATA_cn
```

```python
DATA_data.shape
```

```python
df = DATA_data[DATA_data["ORGAN"].isin(["HR"])]
df = df[df["AGE_GROUP"].isin(["P"])]
df = df[df["PREV_TX"].isin(["N"])]
df = df[df["TXED"].isin(["1", 1])]
df = df[df["GSTATUS"] != "."]
```

```python
df.shape
```

```python
df.duplicated(subset="TRR_ID_CODE").sum()
```

```python
#setting values where the entire cell is ".", "U", "ND", or "I" to NaN
df = df.apply(lambda col: col.map(lambda x: np.nan if isinstance(x, str) and x.strip() == "." else x))

df = df.apply(lambda col: col.map(lambda x: np.nan if isinstance(x, str) and x.strip() == "U" else x))

df = df.apply(lambda col: col.map(lambda x: np.nan if isinstance(x, str) and x.strip() == "ND" else x))

df = df.apply(lambda col: col.map(lambda x: np.nan if isinstance(x, str) and x.strip() == "I" else x))

df = df.apply(lambda col: col.map(lambda x: np.nan if isinstance(x, str) and x.strip() == "C" else x))
```

```python
def try_numeric(col):
    try:
        return pd.to_numeric(col)
    except:
        return col

df = df.apply(try_numeric)
```

```python
df = df.astype({'TX_DATE': 'datetime64[ns]'})
```

```python
cutoff = pd.Timestamp("2016-07-07")
df = df[df["TX_DATE"] >= cutoff]
```
